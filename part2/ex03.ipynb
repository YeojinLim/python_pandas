DataFrame = 여러 개의 Series가 모인 2차원 데이터 구조
Series    = DataFrame에서 하나의 열을 선택한 1차원 데이터 구조

df['열이름'] → Series 반환
df[['열이름']] → DataFrame 반환

```python
import pandas as pd

df = pd.DataFrame({
    "이름": ["김민수", "이서연", "박지훈"],
    "점수": [80, 90, 75]
})

# 열 1개 선택 → Series
s = df["점수"]

display(s)
display(type(s))


# 열 1개를 리스트로 선택 → DataFrame
df2 = df[["점수"]]

display(df2)
display(type(df2))
```

```python
# 1. 실습용 데이터셋

import pandas as pd
import numpy as np

df = pd.DataFrame({
    "직원ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "이름": ["김민준", "이서연", "박지훈", "최유진",
           "정현우", "강수빈", "조민서", "윤도현"],
    "부서": ["개발", "영업", "개발", "인사",
           "영업", "개발", "영업", "인사"],
    "성별": ["남", "여", "남", "여", "남", "여", "여", "남"],
    "매출": [850, 920, np.nan, 780, 1200, 650, 920, np.nan],
    "근무시간": [160, 172, 168, 155, 180, 150, 172, 158],
    "등급": ["B", "A", "B", "C", "A", "C", "A", None],
    "입사년도": [2020, 2019, 2021, 2022, 2018, 2023, 2019, 2022]
})

display(df)
```

2. Series 메서드의 분류

[데이터 탐색]
head → tail → info → describe

[통계]
mean / median / mode
sum / min / max
std / var
count / quantile

[범주형 데이터]
unique → nunique → value_counts

[결측치]
isna / isnull
notna / notnull
      ↓
fillna / dropna

[데이터 변환]
apply / map / replace
where / mask

[시계열·변화]
shift → diff → pct_change

```python
# 3. 실습
# 1) 기본정보

s = df["매출"]
display("s type:",type(s))

display(df["매출"].head(2)) # 기본값 5
display(df["매출"].tail(2))

display(df["매출"].info())
display(df["매출"].describe())
display(df["등급"].describe())
```

```python
# 2) 통계
display("평균:", df["매출"].mean())
display("중앙값:", df["매출"].median())

# 범주타입 - 최빈값은 빈도가 같으면 두개이상을 반환해야 하기때문에 Series타입으로 반환
display("최빈값:", df["등급"].mode())
display("최빈값:", df["성별"].mode())

display("합계:", df["매출"].sum())
display("최대:", df["매출"].max())
display("최소:", df["매출"].min())

display("분산:", df["매출"].var())
display("표준편차:", df["매출"].std())

display("1분위수:", df["매출"].quantile(0.25))
display("2분위수:", df["매출"].quantile(0.50))
display("3분위수:", df["매출"].quantile(0.75))
```

```python
# 데이터 개수
# Series.count()메서드 vs len()함수
display(df["매출"].count()) # NaN 제외
display(len(df["매출"])) # NaN 포함
```

```python
# 3) 고유값 - 범주형 데이터에 주로 사용
display(df["성별"].unique())
display(df["부서"].unique())

# 고유값 개수
display(df["성별"].nunique())
display(df["부서"].nunique())
```

```python
# 4) 빈도 - 범주형 데이터에 주로 사용
display(df["성별"].value_counts())
display(df["부서"].value_counts())
```

```python
# 5) 결측치 - 결측값이면 True
display(df["매출"].isna())
display(df["매출"].isnull())

display(df["매출"].notna()) # not na
display(df["매출"].notnull()) # not null
```

```python
# fillna - 결측값 대체
sales = df["매출"]
# 평균값으로 대체
sales = sales.fillna(sales.mean())
display(sales)
```

```python
# dropna - 결측값 제거
sales = df["매출"]
sales_clean = sales.dropna()
display(sales_clean)
```

```python
# 6) 정렬 - 결측치는 오름/내림 관계없이 마지막
display(df["매출"].sort_values(ascending=False))
display(df["매출"].sort_values())

# 문자값도 정렬
display(df["이름"].sort_values())
# 값이 아닌 인덱스 순으로 정렬
display(df["이름"].sort_index(ascending=False))
```

```python
# 7) 문자열 - str속성 사용
display(df["이름"].str.contains("박"))

names1 = pd.Series([
    " 김민준 ",
    " 이서연",
    "박 지훈 ",
    " 최유진 "
])

names2 = names1.str.strip() # 앞뒤공백제거
display(names2)

names3 = names1.str.replace(" ","") # 공백을 빈문자열로 변경
display(names3)
```

```python
# 8) 형변환
employee_id = df["직원ID"]
print()
print(employee_id.info())
employee_id = employee_id.astype(str)
print()
print(employee_id.info())
```

```python
gender = df["성별"]
print()
print(gender.info())
gender = gender.astype("category")
print()
print(gender.info())
```

```python
# 숫자로 변경
sales = pd.Series(["850", "920", "780", "잘못된값"])
# sales = pd.to_numeric(sales) ERROR
# errors="coerce"는 숫자로 변환할 수 없는 값을 NaN으로 바꾼다.
sales = pd.to_numeric(sales, errors="coerce")
display(sales)

sales = pd.Series(["A", "B", "C"])
sales = pd.to_numeric(sales, errors="coerce")
display(sales)
```

```python
# 9) 조건처리
# where(조건) True -> 원래값, False -> NaN
# 매출이 800 이상인 값은 그대로 두고, 800 미만은 NaN으로 만든다.
result = df["매출"].where(df["매출"] >= 800)
display(result)

# mask() — 조건에 맞는 값 변경
# 매출이 800 미만인 값을 0으로 변경한다.
result = df["매출"].mask(df["매출"] < 800, 0)
display(result)
```

```python
# 10) 중복 여부 확인, 중복 제거
display(df["부서"].duplicated())

unique_department = df["부서"].drop_duplicates()
display(unique_department)
```

```python
# 11) 데이터 값 변경 

# 특정 값만 변경
department_eng1 = df["부서"].replace({
    "개발": "Development"
})
display(department_eng1)


# 전체 시리즈를 변경 - 맵핑되지 않은 값은 NaN으로 변경된다. -> 전체를 변경시에만 사용
department_eng2 = df["부서"].map({
    "개발": "Development"
})
display(department_eng2)

# apply()메서드에 함수를 인자로 전달(고차함수)
# 고차함수 : https://cafe.naver.com/jjdev/19084
def sales_grade(value): # value 시리즈가 아니다. val.isna() 형태는 사용X, pd.isna(값) 함수를 사용 하여야 한다
    if pd.isna(value):
        return "미집계"
    elif value >= 1000:
        return "우수"
    elif value >= 800:
        return "보통"
    else:
        return "낮음"

grade = df["매출"].apply(sales_grade) # 시리즈의 값들에 순차적으로 sales_grade함수를 적용
display(grade)
```

고차변수 참고 https://cafe.naver.com/jjdev/19084

```python
# 12) 누적
display(df["매출"].cumsum()) # NaN제외

display(df["매출"].cummax()) # 슬라이딩 비교 후 큰값만 남기. NaN제외

display(df["매출"].cummin()) # 슬라이딩 비교 후 작은값만 남기. NaN제외

# 누적곱
growth = pd.Series([1.1, 1.2, 0.9, 1.1])
display(growth.cumprod())
```

```python
# 13) 시계열 데이터에 쉬프트 연산

# 값 이동 : 한칸 이동(0번째 값은 NaN으로 대체)
display(df["매출"].shift())
# 값 차이 : 이전값과의 차이(0번째 값은 앞에 값이 없으니 NaN)
display(df["매출"].diff())

# 이전 값 대비 몇% 증가/감소 100->110이면 0.1 10%, 100->90 -0.1, 100->100 0
display(df["매출"].pct_change())
```

```python
# 14) 기타
# 값 포함 isin()
display(df["성별"].isin(["male"]))
display(df["성별"].isin(["남"]))
display(df["성별"].isin(["남", "여"]))

# 무작위 추출 sample()
display(df["이름"].sample(3, random_state=42)) 
# random_state값이 동일하면 동일한 랜덤 알고리즘을 적용 -> 같은값
display(df["이름"].sample(3, random_state=42))
display(df["이름"].sample(3, random_state=0))
```
