1. concat() 함수

# 전처리 - 데이터 결합하기
# concat() 함수 : 같은 구조의 데이터를 합치는 것
# axis=0 → 행을 아래로 붙인다
# axis=1 → 열을 옆으로 붙인다
# SQL의 JOIN과는 다르다.
# concat은 단순히 축을 기준으로 이어 붙이는 것이고, 
# 공통 키를 기준으로(SQL의 JOIN 유사) 데이터를 결합하려면 merge()나 join()을 사용한다.

https://pandas.pydata.org/docs/reference/api/pandas.concat.html

pd.concat(
    objs,
    axis=0,
    join="outer",
    ignore_index=False,
    keys=None
)

pd.concat(
    objs,
    axis=0,
    join="outer",
    ignore_index=False,
    keys=None
)

주요 매개변수
- objs : 결합할 객체 -> [df1, df2]
- axis : 결합 방향   -> 0,1
- join : 다른 축의 처리 방법 -> "outer", "inner" 
(join은 “공통 부분만 붙일 것인가, 전체를 유지할 것인가”를 결정하는 옵션이다. SQL문의 JOIN같은 것이 아니다.)
- ignore_index : 기존 인덱스 무시  -> True, False
- keys : 결합된 데이터에 계층적 인덱스 부여  -> ["A", "B"]

```python
import pandas as pd

df1 = pd.DataFrame({
    "이름": ["홍길동", "김철수"],
    "점수": [80, 90]
})

df2 = pd.DataFrame({
    "이름": ["이영희", "박민수"],
    "점수": [85, 95]
})

display(df1)
display(df2)
```

```python
# ignore_index=False : 기본값
# axis=0 위아래 결합(기본값) -> 행추가

df = pd.concat([df1, df2], ignore_index=False, axis=0) # ignore_index 기본값 False
display(df) # 결과 : 인덱스 유지
```

```python
# ignore_index=True 

df = pd.concat([df1, df2], ignore_index=True, axis=0)
display(df) # 결과 : 인덱스 리셋
```

```python
# axis=1 : 생략하면 기본값 0
# axis=0 좌우 결합 -> 열추가

df = pd.concat([df1, df2], axis=1) # axis=0 행축, axix=1 열축 기본값 0
display(df)
```

```python
# join="outer" vs join="inner"
# join은 “공통 부분만 붙일 것인가, 전체를 유지할 것인가”를 결정

df1 = pd.DataFrame({
    "이름": ["김민수", "이서연"],
    "점수": [80, 90]
})

df2 = pd.DataFrame({
    "이름": ["박지훈", "최유진"],
    "등급": ["A", "B"]
})
display(df1)
display(df2)
```

```python
# 두 데이터 프레임의 공통된 열은 "이름"열만 존재한다.

df = pd.concat([df1, df2], join="outer", ignore_index=True) # join생략하면 기본값이 outer
display(df)
```

```python
df = pd.concat([df1, df2], join="inner", ignore_index=True)
display(df)
```

```python
# axis=1

import pandas as pd

df = pd.DataFrame({
    "이름": ["홍길동", "김철수", "이영희", "박민수"],
    "점수": [90, 80, 95, 70]
})

grade = pd.DataFrame({
    "등급": ["A", "B", "A", "C"]
})

result = pd.concat([df, grade], axis=1)

display(result)
```

```python
# # 열의 개수가 다르면 -> join="outer"이면(기본값)  NaN으로 채워진다
df = pd.DataFrame({
    "이름": ["홍길동", "김철수", "이영희", "박민수"],
    "점수": [90, 80, 95, 70]
})

grade2 = pd.DataFrame({
    "등급": ["A", "B", "A"]
})

result = pd.concat([df, grade2], axis=1)

display(result)
```

```python
pd.concat([df, grade2], axis=1, join="inner")
```

2. merge() 함수

# merge()는 두 DataFrame의 공통된 열(키)을 기준으로 데이터를 결합할 때 사용하는 함수
# SQL의 JOIN문의 경험이 있다면 merge()는 이해하기 쉽다.(JOIN 개념 자체는 동일하다)

https://pandas.pydata.org/docs/reference/api/pandas.merge.html

pd.merge(
    left,
    right,
    how="inner",
    on=None,
    left_on=None,
    right_on=None,
    suffixes=("_x", "_y")
)

주요 매개변수

left : 왼쪽 DataFrame -> df1
right : 오른쪽 DataFrame -> df2
how : 결합 방법 -> "inner", "left", "right", "outer"
on : 양쪽에 공통으로 존재하는 결합 기준 열 -> "고객ID" (key)
left_on : 왼쪽 DataFrame의 기준 열 -> "고객번호"
right_on : 오른쪽 DataFrame의 기준 열 -> "회원번호"
suffixes : 같은 이름의 열 이름에 붙일 접미사 -> ("_x", "_y")

```python
import pandas as pd

df_customer = pd.DataFrame({
    "고객ID": [101, 102, 103, 104],
    "이름": ["홍길동", "김철수", "이영희", "박민수"]
})

df_score = pd.DataFrame({
    "고객ID": [101, 102, 103, 105],
    "점수": [90, 80, 95, 70]
})


display(df_customer)
display(df_score)

result = pd.merge(
    df_customer,
    df_score,
    on="고객ID",
    how="inner" # inner(기본값), left, right, outer
)

display(result)
```

```python
result = pd.merge(
    df_customer,
    df_score,
    on="고객ID",
    how="left"
)

display(result)
```

```python
result = pd.merge(
    df_customer,
    df_score,
    on="고객ID",
    how="right"
)

display(result)
```

```python
result = pd.merge(
    df_customer,
    df_score,
    on="고객ID",
    how="outer"
)

display(result)
```
