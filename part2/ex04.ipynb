< DataFrame 다루기 - 인덱스(label) 관리 >

```python
# 인덱스 조작(레이블 인덱스. 위치값 아니다)

# 현재 인덱스를 확인
import pandas as pd

df = pd.DataFrame({
    "이름": ["김민준", "이서연", "박지훈"],
    "점수": [85, 90, 75]
})

print(df)

print(df.index)
```

```python
print(df.index.tolist()) # 레이블 인덱스 값을 리스트로 반환
print(df.index.name) # 레이블 인덱스 이름, 없으면 None
```

```python
# 인덱스 이름 설정
df.index.name = "학생구분"
print(df)
```

```python
# 인덱스 변경
df.index = ["학생1", "학생2", "학생3"]

print(df)
```

```python
# set_index : 열 -> 인덱스
# 열을 인덱스로 변경

df = pd.DataFrame({
    "학생ID": [101, 102, 103],
    "이름": ["김민준", "이서연", "박지훈"],
    "점수": [85, 90, 75]
})

print(df)

df2 = df.set_index("학생ID") # 변경된 데이터프레임을 반환

print(df2)

# 반환없이 원본의 인덱스를 변경  inplace=True 매개값 추가
df.set_index("학생ID", inplace=True) # 컬럼이름이 인덱스이름이 된다.
df
```

```python
# reset_index : 인데스 -> 열

# 인덱스를 열로 변경
df = pd.DataFrame({
    "학생ID": [101, 102, 103],
    "이름": ["김민준", "이서연", "박지훈"],
    "점수": [85, 90, 75]
})

print(df)

df2 = df.reset_index(names="번호") # names생략시 인덱스이름이 기본값, 이름이 없으면 "index"

print(df2)
```

```python
# 기존 인덱스를 삭제

df = pd.DataFrame({
    "이름": ["김민준", "이서연", "박지훈"]
}, index=[10, 20, 30])

print(df)

df2 = df.reset_index(drop=True) # reset_index의 기본값은 인덱스를 열로 이동인데, drop=True이면 삭제
print(df2)
```

```python
# 인덱스 정렬
df = pd.DataFrame({
    "이름": ["김민준", "이서연", "박지훈"]
}, index=[30, 10, 20])

print(df)

df2 = df.sort_index()

print(df2)
```
