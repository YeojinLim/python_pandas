```python
# 2. 워드 클라우드 시각화
# 텍스트 데이터에서 어떤 단어가 많이 등장하는지 빠르게 탐색하는 도구
```

```python
text = " ".join(df["opinion"])
print(text)
```

```python
# 수치값으로 먼저 탐색
from collections import Counter

# 수치값으로 확인
words = text.lower().split()
print(words)
word_count = Counter(words)
word_count
```

```python
# 시각화
from wordcloud import WordCloud
import matplotlib.pyplot as plt

wordcloud = WordCloud(
    width=800,
    height=500,
    background_color="white",
    stopwords=set()       # 기본 불용어 사용 안 함
).generate(text)

plt.figure(figsize=(10, 6))
plt.imshow(wordcloud)
plt.axis("off")
plt.show()
```

```python
# 분석에는 큰 의미가 없는 단어(불용어) 제외

stopwords = {
    "the", "and", "is", "to", "was",
    "a", "of", "in", "with", "me",
    "my", "more", "than", "I"
}

wordcloud = WordCloud(
    width=800,
    height=500,
    background_color="white",
    stopwords=stopwords # 불용어 설정
).generate(text)

plt.figure(figsize=(10, 6))
plt.imshow(wordcloud)
plt.axis("off")
plt.show()
```
