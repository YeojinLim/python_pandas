< 판다스 DataFrame 입력, 수정, 삭제와 통계관련 메서드 >

```python
# 1. 입력, 수정, 삭제

import pandas as pd

df = pd.DataFrame({
    '이름': ['김철수', '이영희', '박민수', '최수진', '정우진'],
    '학년': [1, 2, 1, 3, 2],
    '점수': [85, 92, 78, 95, 88],
    '출석': [90, 95, 80, 98, 85]
})

df
```

```python
# 새로운 열 추가

df['성적'] = ['B', 'A', 'C', 'A+', 'B+']

df
```

```python
# 계산해서 새로운 열 만들기

df['총점'] = df['점수'] + df['출석']

df
```

```python
# 특정 값 수정

df.loc[0, '점수'] = 90

df
```

```python
df.loc[df['점수'] < 80, '점수'] = 80

df
```

삭제 drop(열이름 or 레이블 인덱스)메서드

* drop()에 매개값으로 위치값을 사용할 수 없다
- 특정 위치 값들을 삭제 X -> 값을 공백, NaN, ... , 결측치 변경 O
​- 행들을 선택해서 삭제 - 세로축으로 선택 : axis = 0 행
- 열들을 선택해서 삭제 - 가로축으로 선택 : axis = 1 열

```python
df = df.drop(0, axis=0) # axis는 생략하면 0

df
```

```python
# - 열삭제 : drop('열 레이블', axis=1)
# - axis=1은 열축(가로방향 - 하나이상의 열을 삭제 가능)
# '총점'을 열 방향(axis=1)에서 삭제

df.drop('총점', axis=1) # axis는 생략하면 0

df
```
