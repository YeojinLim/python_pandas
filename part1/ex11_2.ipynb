2. 앤스콤 데이터

앤스콤의 사중주(Anscombe's quartet)는 평균, 분산, 상관계수, 회귀선이 거의 같은 4개의 데이터셋이지만  실제 데이터 분포와 그래프 모양은 크게 다르다는 것을 보여주는 대표적인 예제다.

```python
import matplotlib.pyplot as plt


x1 = [10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5]
y1 = [8.04, 6.95, 7.58, 8.81, 8.33, 9.96, 7.24, 4.26, 10.84, 4.82, 5.68]

x2 = [10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5]
y2 = [9.14, 8.14, 8.74, 8.77, 9.26, 8.10, 6.13, 3.10, 9.13, 7.26, 4.74]

x3 = [10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5]
y3 = [7.46, 6.77, 12.74, 7.11, 7.81, 8.84, 6.08, 5.39, 8.15, 6.42, 5.73]

x4 = [8, 8, 8, 8, 8, 8, 8, 19, 8, 8, 8]
y4 = [6.58, 5.76, 7.71, 8.84, 8.47, 7.04, 5.25, 12.50, 5.56, 7.91, 6.89]
```

```python
import numpy as np

# 4개의 데이터셋을 하나로 묶기
datasets = [
    (x1, y1),
    (x2, y2),
    (x3, y3),
    (x4, y4)
]


# 데이터셋별 통계량 계산
for i, (x, y) in enumerate(datasets, start=1):

    # 리스트를 NumPy 배열로 변환
    x = np.array(x)
    y = np.array(y)

    # 평균
    x_mean = np.mean(x)
    y_mean = np.mean(y)

    # 분산
    # ddof=1 → 표본분산
    x_var = np.var(x, ddof=1)
    y_var = np.var(y, ddof=1)

    # 상관계수
    correlation = np.corrcoef(x, y)[0, 1]

    # 회귀선
    # y = a*x + b
    # a = 기울기, b = 절편
    a, b = np.polyfit(x, y, 1)

    print(f"\n[Dataset {i}]")

    print(f"x 평균     : {x_mean:.2f}")
    print(f"y 평균     : {y_mean:.2f}")

    print(f"x 분산     : {x_var:.2f}")
    print(f"y 분산     : {y_var:.2f}")

    print(f"상관계수   : {correlation:.2f}")

    print(f"회귀선 기울기 : {a:.2f}")
    print(f"회귀선 절편   : {b:.2f}")

    print(f"회귀식     : y = {a:.2f}x + {b:.2f}")
```

```python
import matplotlib.pyplot as plt


# ============================================================
# 1. 전체 Figure 생성
# ============================================================

# 전체 그래프를 담을 큰 도화지(Figure)를 생성한다.
fig = plt.figure(figsize=(10, 8))


# ============================================================
# 2. 첫 번째 그래프 공간 생성
# ============================================================

# 2행 2열로 나누고 그중 첫 번째 위치에
# 그래프 공간(Axes)을 생성한다.
axes1 = fig.add_subplot(2, 2, 1)

# 첫 번째 데이터셋을 산점도로 그린다.
axes1.scatter(x1, y1)

# 그래프 제목을 설정한다.
axes1.set_title("Dataset 1")


# ============================================================
# 3. 두 번째 그래프 공간 생성
# ============================================================

# 2행 2열 중 두 번째 위치에
# 그래프 공간(Axes)을 생성한다.
axes2 = fig.add_subplot(2, 2, 2)

# 두 번째 데이터셋을 산점도로 그린다.
axes2.scatter(x2, y2)

# 그래프 제목을 설정한다.
axes2.set_title("Dataset 2")


# ============================================================
# 4. 세 번째 그래프 공간 생성
# ============================================================

# 2행 2열 중 세 번째 위치에
# 그래프 공간(Axes)을 생성한다.
axes3 = fig.add_subplot(2, 2, 3)

# 세 번째 데이터셋을 산점도로 그린다.
axes3.scatter(x3, y3)

# 그래프 제목을 설정한다.
axes3.set_title("Dataset 3")


# ============================================================
# 5. 네 번째 그래프 공간 생성
# ============================================================

# 2행 2열 중 네 번째 위치에
# 그래프 공간(Axes)을 생성한다.
axes4 = fig.add_subplot(2, 2, 4)

# 네 번째 데이터셋을 산점도로 그린다.
axes4.scatter(x4, y4)

# 그래프 제목을 설정한다.
axes4.set_title("Dataset 4")


# ============================================================
# 6. 그래프 간격 조정
# ============================================================

# 각 그래프의 제목 등이 서로 겹치지 않도록
# 그래프 사이의 간격을 자동으로 조정한다.
fig.tight_layout()


# ============================================================
# 7. Figure 출력
# ============================================================

# 최종적으로 Figure를 화면에 출력한다.
plt.show()
```
