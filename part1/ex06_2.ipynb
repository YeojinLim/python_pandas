```python
import pandas as pd

df = pd.DataFrame({
    '이름': ['김철수', '이영희', '박민수', '최수진', '정우진'],
    '학년': [1, 2, 1, 3, 2],
    '점수': [85, 92, 78, 95, 88],
    '출석': [90, 95, 80, 98, 85]
})
```

```python
# 2. 통계 관련 메서드
# 1) 전체 통계량

# 데이터 개수
print(df.count())
```

```python
# 합계
print(df.sum(numeric_only=True))
```

```python
# 평균
print(df.mean(numeric_only=True))
```

```python
print(df.median(numeric_only=True))
```

```python
#최소값
print(df.min(numeric_only=True))
```

```python
#최대값
print(df.max(numeric_only=True))
```

```python
print(df.std(numeric_only=True))
```

```python
print(df.std(numeric_only=True))
```

```python
print(df['점수'].max() - df['점수'].min())
```

```python
print(df.describe())
```

```python
# 2) 특정 열의 통계량
print(df['점수'].mean())
print(df['점수'].median())
print(df['점수'].min())
print(df['점수'].max())
print(df['점수'].std())
```

```python
# 3) 기타

print(df.isnull())
# True  → 결측값
# False → 결측값 아님
```

```python
print(df.isnull().sum()) # 결측값 개수
```

```python
print(df.duplicated()) # 중복 행 여부
# True  → 중복 행
# False → 중복 아님
```

```python
print(df['학년'].unique()) # 고유값 확인 : 학년 열에 어떤 값들이 존재하는지 확인한다.
print(df['학년'].nunique()) # 고유값 개수
```

```python
print(df['학년'].value_counts()) # 값별 개수
```

```python
print(df['점수'].isnull()) # 특정 열의 결측값
```

```python
# 학년의 데이터 타입을 category 타입(범주형)으로 변환한 결과를 
# 다시 df['학년']에 저장하여 df의 학년 데이터 타입을 변경
df['학년'] = df['학년'].astype('category')

print(df['학년'].dtype)
```
